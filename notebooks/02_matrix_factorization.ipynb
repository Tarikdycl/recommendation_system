```python
import numpy as np

R = np.array([
    [5,4,0],
    [4,0,1]
], dtype=float)
# R is actual observations

U = np.array([
    [0.5,0.3],
    [0.4,0.7]
], dtype=float)

# U is hidden features from users we are gonna update them; use them as parameters

V = np.array([
    [0.6,0.2],
    [0.8,0.1],
    [0.2,0.9]
])

# V is hidden features from movie (sci-fi, romantic) we are gonna use this as parameters as well
```

```python
predictions = U @ V.T
print(f"Actual ratings {R}")

print(f"\nPredictions: {predictions}")
```

```python
mask = R != 0

print(mask)
```

```python
errors = R[mask] - predictions[mask]
print(errors)
```

```python
mse = np.mean(errors ** 2)

print("MSE :", mse)
```

```python
actual = R[0, 0]      # 5
prediction = predictions[0, 0]  # 0.36
error = actual - prediction
```

```python
user_vec = U[0].copy()
movie_vec = V[0].copy()

print(user_vec,movie_vec)
```

```python
lr = 0.01
```

```python
U[0]+= lr * error * movie_vec
V[0]+= lr * error * user_vec

new_predictions = U[0] @ V[0]

print(f"Old prediction: {prediction}\nNew prediction: {new_predictions}")
```

```python
epoch = 1000

for epochs in range(epoch):
    for u in range(R.shape[0]): # Users
        for i in range(R.shape[1]): #Movies
            if R[u,i] == 0:
                continue
            prediction = U[u] @ V[i]
            error = R[u,i] - prediction

            user_vec = U[u].copy()

            U[u] += lr*error*V[i]
            V[i] += lr*error*user_vec
        
```

```python
predictions = U @ V.T

print(f"Actual ratings: {R}\nPredicted Ratings: {predictions}")
```

```python
mask = R != 0

errors = R[mask] - predictions[mask]
mse = np.mean(errors ** 2)

print("\nFinal MSE:", mse)
```

```python
R = np.array([
    [5, 4, 1, 0, 0, 3],
    [5, 4, 2, 5, 4, 0],
    [1, 2, 5, 1, 0, 4],
    [4, 5, 1, 4, 5, 0],
    [0, 4, 2, 5, 4, 3],
    [2, 1, 5, 0, 1, 4]
], dtype=float)

num_users = R.shape[0]
num_movies = R.shape[1]
num_features = 3

np.random.seed(42)

U = np.random.rand(num_users, num_features)
V = np.random.rand(num_movies, num_features)

print("R shape:", R.shape)
print("U shape:", U.shape)
print("V shape:", V.shape)

print("\nU:")
print(U)

print("\nV:")
print(V)
```

```python
R_train = R.copy()

test_cases = [
    (0, 1),  # User 0, Movie 1
    (1, 3),  # User 1, Movie 3
    (2, 2),  # User 2, Movie 2
    (3, 4)   # User 3, Movie 4
]

test_actuals = []

for u, i in test_cases:
    test_actuals.append(R[u, i])
    R_train[u, i] = 0

print("R_train:")
print(R_train)

print("\nHidden test ratings:")
print(test_actuals)
```

```python
# Reinitalizing U and V for training
```

```python
np.random.seed(42)

U = np.random.rand(num_users, num_features)
V = np.random.rand(num_movies, num_features)
```

```python
lr = 0.01
epochs = 1000

for epoch in range(epochs):

    for u in range(R_train.shape[0]):
        for i in range(R_train.shape[1]):

            if R_train[u, i] == 0:
                continue

            prediction = U[u] @ V[i]
            error = R_train[u, i] - prediction

            user_vec = U[u].copy()
            movie_vec = V[i].copy()

            U[u] += lr * error * movie_vec
            V[i] += lr * error * user_vec
```

```python
test_predictions = []

for u, i in test_cases:
    predicted = U[u] @ V[i]
    test_predictions.append(predicted)

    print(
        f"User {u}, Movie {i} | "
        f"Actual: {R[u, i]:.2f} | "
        f"Predicted: {predicted:.2f}"
    )
```

```python
test_actuals = np.array(test_actuals)
test_predictions = np.array(test_predictions)

rmse = np.sqrt(
    np.mean(
        (test_actuals - test_predictions) ** 2
    )
)

print("\nTest RMSE:", rmse)
```

```python
# Adding L2 regularization error
```

```python
lr = 0.01
epochs = 1000
reg = 0.02

for epoch in range(epochs):

    for u in range(R_train.shape[0]):
        for i in range(R_train.shape[1]):

            if R_train[u, i] == 0:
                continue

            prediction = U[u] @ V[i]
            error = R_train[u, i] - prediction

            user_vec = U[u].copy()
            movie_vec = V[i].copy()

            U[u] += lr * (
                error * movie_vec
                - reg * user_vec
            )

            V[i] += lr * (
                error * user_vec
                - reg * movie_vec
            )
```

```python
test_predictions = []

for u, i in test_cases:
    predicted = U[u] @ V[i]
    test_predictions.append(predicted)

    print(
        f"User {u}, Movie {i} | "
        f"Actual: {R[u, i]:.2f} | "
        f"Predicted: {predicted:.2f}"
    )
```

```python
test_actuals = np.array(test_actuals)
test_predictions = np.array(test_predictions)

rmse = np.sqrt(
    np.mean(
        (test_actuals - test_predictions) ** 2
    )
)

print("Regularized Test RMSE:", rmse)
```

```python
known_ratings = R_train[R_train != 0]

mu = np.mean(known_ratings)

user_bias = np.zeros(num_users)
movie_bias = np.zeros(num_movies)

np.random.seed(42)

U = np.random.rand(num_users, num_features)
V = np.random.rand(num_movies, num_features)

print("Global mean:", mu)
```

```python
lr = 0.01
epochs = 1000
reg = 0.02

for epoch in range(epochs):

    for u in range(R_train.shape[0]):
        for i in range(R_train.shape[1]):

            if R_train[u, i] == 0:
                continue

            prediction = (
                mu
                + user_bias[u]
                + movie_bias[i]
                + U[u] @ V[i]
            )

            error = R_train[u, i] - prediction

            user_vec = U[u].copy()
            movie_vec = V[i].copy()

            # update biases
            user_bias[u] += lr * (
                error - reg * user_bias[u]
            )

            movie_bias[i] += lr * (
                error - reg * movie_bias[i]
            )

            # update latent factors
            U[u] += lr * (
                error * movie_vec
                - reg * user_vec
            )

            V[i] += lr * (
                error * user_vec
                - reg * movie_vec
            )
```

```python
test_predictions = []

for u, i in test_cases:

    predicted = (
        mu
        + user_bias[u]
        + movie_bias[i]
        + U[u] @ V[i]
    )

    predicted = np.clip(predicted, 1, 5)

    test_predictions.append(predicted)

    print(
        f"User {u}, Movie {i} | "
        f"Actual: {R[u, i]:.2f} | "
        f"Predicted: {predicted:.2f}"
    )
```

```python
test_predictions = np.array(test_predictions)

rmse = np.sqrt(
    np.mean(
        (test_actuals - test_predictions) ** 2
    )
)

print("\nBias + Regularization Test RMSE:", rmse)
```

```python
results = []

for num_features in [2, 3, 4]:
    for reg in [0.001, 0.01, 0.02, 0.05]:

        np.random.seed(42)

        U = np.random.rand(num_users, num_features)
        V = np.random.rand(num_movies, num_features)

        user_bias = np.zeros(num_users)
        movie_bias = np.zeros(num_movies)

        mu = np.mean(R_train[R_train != 0])

        lr = 0.01
        epochs = 1000

        for epoch in range(epochs):

            for u in range(R_train.shape[0]):
                for i in range(R_train.shape[1]):

                    if R_train[u, i] == 0:
                        continue

                    prediction = (
                        mu
                        + user_bias[u]
                        + movie_bias[i]
                        + U[u] @ V[i]
                    )

                    error = R_train[u, i] - prediction

                    user_vec = U[u].copy()
                    movie_vec = V[i].copy()

                    user_bias[u] += lr * (
                        error - reg * user_bias[u]
                    )

                    movie_bias[i] += lr * (
                        error - reg * movie_bias[i]
                    )

                    U[u] += lr * (
                        error * movie_vec
                        - reg * user_vec
                    )

                    V[i] += lr * (
                        error * user_vec
                        - reg * movie_vec
                    )

        preds = []

        for u, i in test_cases:
            pred = (
                mu
                + user_bias[u]
                + movie_bias[i]
                + U[u] @ V[i]
            )

            pred = np.clip(pred, 1, 5)
            preds.append(pred)

        preds = np.array(preds)

        rmse = np.sqrt(
            np.mean((test_actuals - preds) ** 2)
        )

        results.append(
            (num_features, reg, rmse)
        )
```

```python
results = sorted(
    results,
    key=lambda x: x[2]
)

for result in results:
    print(result)
```
